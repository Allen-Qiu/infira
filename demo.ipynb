# InFiRA: An LLM-based agent for intelligent financial report analysis
Owing to GPU rental expenses, server will be started on request! qjt163@163.com


```python
import requests

# remote server
url="http://infira.cpolar.cn/qr"
llm = "ds"

# code of the company
report = "000732.pl"

def run(question):
    params = {"query": question, "model":llm, "report":report}
    response = requests.get(url, params=params)
    return response.text.strip()
```

### Question 1: Who is chairman of this company

```python
s = "该公司董事长是谁？"
response = run(s)
print(response)
s = f"llm:please translate the following contents to English\n{response}"
response = run(s)
print(response)
```
